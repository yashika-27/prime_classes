```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
tips=sns.load_dataset('tips')
print(type(tips))
tips.head()
```

```python
sns.set_theme(style='white')
sns.relplot(

    data=tips,
    x='total_bill',
    y='tip',
    hue='smoker',
    size='size',
    style='smoker',
    row='time',
    col='sex'
)
```

```python
flights=sns.load_dataset('flights')
```

```python
sns.lineplot(
    data=flights,
    x='year',
    y='passengers',
    errorbar=('ci',False)
)
```

```python
sns.barplot(
    data=tips,
    x='day',
    y='tip',
    hue='sex'
)
```

```python
sns.boxplot(
    data=tips,
    x='day',
    y='tip',
    hue='sex'
)
```

```python
penguins = sns.load_dataset("penguins")
sns.histplot(
    data=penguins,
    x='body_mass_g',
    bins=30
)
```

```python
flight_pivot=flights.pivot(index='month',columns='year',values='passengers')
sns.heatmap(
    data=flight_pivot,
    cmap='coolwarm',
    annot=True,
    fmt='d'
)
plt.title('flight passengers over years')
plt.tight_layout()
```
