```python
def is_denied(row):
    return row['pc_ClaimStatus'] == 4
```

```python
def check_timely_filing(row):
    days = (row['pc_ReceivedDate'] - row['ec_ServiceDateFrom']).days
    limit = 365 if row['ec_InsuranceType'] == "Medicare" else 180

    if days > limit:
        return "Not Recoverable", {
            "days_delay": days,
            "limit": limit,
            "status": "exceeds limit"
        }
    else:
        return "Recoverable", {
            "days_delay": days,
            "limit": limit,
            "status": "within limit"
        }
```

```python
def check_missing_info(row):
    auth = row['ec_PriorAuthorization']

    if pd.isna(auth) or str(auth).strip() == "":
        return "Needs Review", {
            "issue": "Missing prior authorization",
            "action": "Check if retro authorization possible"
        }

    return "Needs Review", {
        "issue": "Other missing information"
    }
```

```python
def check_duplicate(row, df):
    duplicates = df[
        (df['ec_ServiceDateFrom'] == row['ec_ServiceDateFrom']) &
        (df['cd_ProcedureCode'] == row['cd_ProcedureCode']) &
        (df['pc_ClaimID'] != row['pc_ClaimID'])
    ]

    return "Needs Review", {
        "duplicate_count": len(duplicates),
        "note": "No similar claim found in dataset"
    }
```

```python
def check_medical(row):
    return "Needs Review", {
        "note": "Requires clinical validation"
    }
```

```python
def analyze_claim(row, df):

    if row['pc_ClaimStatus'] != 4:
        return "Paid", "No Action Needed", {}, None

    carc = str(row['pcla_AdjustmentReason'])

    # 🔴 CARC 29 (Late Filing)
    if carc == "29":
        decision, evidence = check_timely_filing(row)

        validity = True if evidence["status"] == "exceeds limit" else False

        return "Claim denied for late filing", decision, evidence, validity

    # 🟡 CARC 16 (Missing Info)
    elif carc == "16":
        decision, evidence = check_missing_info(row)

        rate = row.get('denial_rate', 0)

        if rate < 0.3:
            decision = "Recoverable"
            evidence["pattern_insight"] = "Missing info rarely denied → likely recoverable"

        elif rate > 0.7:
            decision = "Not Recoverable"
            evidence["pattern_insight"] = "Frequently denied → likely valid denial"

        else:
            evidence["pattern_insight"] = "Moderate denial pattern"

        return "Claim denied due to missing information", decision, evidence, None

    # 🟡 CARC 18 (Duplicate)
    elif carc == "18":
        decision, evidence = check_duplicate(row, df)

        rate = row.get('carc_denial_rate', 0)

        if rate < 0.2:
            evidence["pattern_insight"] = "Duplicates are rare → possible incorrect denial"

        elif rate > 0.6:
            evidence["pattern_insight"] = "Duplicates common → likely valid"

        else:
            evidence["pattern_insight"] = "Moderate duplicate occurrence"

        return "Claim denied as duplicate", decision, evidence, None

    # 🟡 CARC 50 (Medical)
    elif carc == "50":
        decision, evidence = check_medical(row)

        rate = row.get('carc_denial_rate', 0)

        if rate < 0.3:
            decision = "Recoverable"
            evidence["pattern_insight"] = "Low denial rate → possible incorrect denial"

        elif rate > 0.7:
            decision = "Not Recoverable"
            evidence["pattern_insight"] = "High denial rate → likely valid denial"

        else:
            evidence["pattern_insight"] = "Moderate denial rate"

        return "Claim denied for medical necessity", decision, evidence, None

    # 🔵 Default
    else:
        return "Unknown", "Needs Review", {}, None
```

```python
def get_confidence(decision, validity):
    if decision == "Not Recoverable":
        return 0.95
    elif decision == "Recoverable":
        return 0.9 if validity == False else 0.85
    else:
        return 0.7
```

```python
def create_output(row, df):
    root_cause, decision, evidence ,validity = analyze_claim(row, df)

    confidence = get_confidence(decision,validity)

    return {
        "claim_id": row['pc_ClaimID'],
        "carc_code": row['pcla_AdjustmentReason'] if row['pcla_AdjustmentReason'] else None,
        "root_cause": root_cause,
        "recoverability": decision,
        "is_denial_valid": validity,
        "evidence": evidence,
        "confidence": confidence
    }
```

```python
features = df[['pcla_AdjustmentAmount', 'denial_rate']]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=42)
df['cluster'] = kmeans.fit_predict(scaled_features)
```

```python
def label_cluster(row):
    amount = row['pcla_AdjustmentAmount']
    rate = row['denial_rate']

    if amount > 8000 and rate < 0.3:
        return "High Value - Low Risk"

    elif amount > 8000 and rate <= 0.6:
        return "High Value - Medium Risk (Focus)"

    elif rate > 0.8:
        return "High Risk - Likely Not Recoverable"

    elif amount < 1000:
        return "Low Value - Low Priority"

    else:
        return "Medium Priority"
```
