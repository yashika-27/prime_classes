```python
import json
```

```python
def load_data(filename):
    with open(filename,'r') as f:
        data=json.load(f)
        return data
```

```python
data = load_data('edge_cases.json')
print(data)
```

```python
def analytical_data(data):
    text_to_num={'one':1,'two':2,'three':3,'four':4,'five':5}
    for user in data:
        # raw_rating=str(user.get('rating')).strip().lower()  .get gives none value and that will convert to None-->'none'
        raw_rating=user.get('rating')
        if (raw_rating==None or str(raw_rating).strip()=="" or str(raw_rating).strip().lower()=='none'):
            user['rating']=None
            continue  
            #One-line mental rule (remember this) =“Handling invalid data must be followed by continue or return.”
            #“We must handle both NoneType values and string placeholders like 'none' before numerical conversion to avoid runtime errors.”
        key=str(raw_rating).strip().lower()
        if(key in text_to_num):
            raw_rating=text_to_num[key]
        user['rating']=raw_rating
        if(float(user['rating'])>5.0):
            user['rating']=None
    return data
            
```

```python
data=analytical_data(data)
print(data)
```
