```python
import json
```

```python
def load_data(fileName):
    with open(fileName,'r') as f:
        data=json.load(f)
        return data
```

```python
data=load_data('amazon_store_data.json')
```

```python
def clean_data(data):
    unique_user= []
    duplicate_user=set() # because set has unique values so its time complexity is less that is O(1)
    # unique_data=set(data)  cant do this as mutables cant be element of set
    text_to_num={'one':1,'two':2,'three':3,'four':4,'five':5}
    for user in data:
        raw_rating=str(user['rating']).strip().lower()
        if raw_rating in text_to_num:
            raw_rating=int(text_to_num[raw_rating])
        user['rating']=raw_rating
        if (user.get('age')==None):
            user['age']=None
        
        if (user['name'] in duplicate_user):
            continue
        else:
            duplicate_user.add(user['name'])
            unique_user.append(user)
    return unique_user
            
   
            
    
```

```python
data=clean_data(data)
```

```python
def get_insights(data):
    total_ratings=0
    poor_rating=0
    for user in data:
        total_ratings+=float(user['rating'])
        if(float(user['rating'])<3.0):
            poor_rating+=1
    print(f"avg of ratings is : {total_ratings/len(data)}")
    print(f"percentage of poor rating : {poor_rating/len(data)*100}%")

    
    
```

```python
get_insights(data)
```

```python
def recommendation(data):
    all_recommends=[]
    for user in data:
        recommends={}
        recommends['name']=user['name']
        if(float(user['rating'])>=4.0):
            recommends['brand']='Apple'
        else:
            recommends['brand']='Samsung'
        all_recommends.append(recommends)
    return all_recommends
    
```

```python
recommendations_per_user=recommendation(data)
print(recommendations_per_user)
```
