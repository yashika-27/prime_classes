```python
import pandas as pd
```

```python
df=pd.read_csv('IRIS.csv')
df
df.head(10)
df.loc[0:10]
df.shape
type(df)
df.info
df.describe()
```

```python
df[(df['petal_length']> 4.5) & (df['species']=='Iris-virginica')]
name='Iris-virginica'
length=4.5
df.query('petal_length > @length & species==@name') #only varables work here not any direct value
```

```python
df.groupby('species').agg({
    'sepal_length':'mean',    #make sure to use comma
    'petal_width':'max',
    'sepal_width':'std'
})
```

```python
df=df.assign(petal_ratio=df['petal_length']/df['petal_width'])
df.groupby('species')['petal_ratio'].mean()
```
