```python
import json
```

```python
def load_data(fileName):
    with open(fileName,'r') as f:
        data=json.load(f)
        return data
```

```python
data=load_data('messy_rating.json')
print(data)
```

```python
def clean_data(data):
    seems_name=set()
    unique_data=[]
    text_to_num={'one':1,'two':2,'three':3,'four':4,'five':5}
    for user in data:
        raw_rating=str(user['rating']).strip().lower()
        if (raw_rating in text_to_num):
            raw_rating=text_to_num[raw_rating]
        user['rating']=raw_rating
        if(user.get('age')==None):
            user['age']=None
        if(user['name'] in seems_name):
            continue
        else:
            seems_name.add(user['name'])
            unique_data.append(user)
    return unique_data
        
```

```python
data=clean_data(data)
print(data)
```

```python
def get_insights(data):
    total_rating=0
    poor_rating=0
    for user in data:
        total_rating+=float(user['rating'])
        if (float(user['rating'])<3.0):
            poor_rating+=1
    print(f"average rating is : {total_rating/len(data)}")
    print(f"poor rating percentage : {poor_rating/len(data)*100}%")
```

```python
get_insights(data)
```

```python
def recommendation(data):
    recommeds=[]
    for user in data:
        curr_recommend={}
        curr_recommend['name']=user['name']
        if(float(user['rating'])>=4.0):
            curr_recommend['brand']='Android'
        else:
            curr_recommend['brand']='IOS'
        recommeds.append(curr_recommend)
    return recommeds
```

```python
recommendations=recommendation(data)
print(recommendations)
```
