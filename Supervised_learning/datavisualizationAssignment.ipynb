```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
sns.set_theme()
```

```python
np.random.seed(0)
scores=np.random.normal(70,10,200)
# fig,axes=plt.subplot()   
# sns.histplot(
#     data=scores,          sns takes dataframes
#     x='scores',
#     bins=20
# )

plt.hist(
    x=scores,
    bins=20
)
plt.axvline(
    x=scores.mean(),
    linewidth=2,
    color='black',
    linestyle='--'
    
)
```

```python
penguins=sns.load_dataset('penguins')
penguins.head()
sns.relplot(
    data=penguins,
    x='body_mass_g',
    y='flipper_length_mm',
    hue='species',
)
```

```python
tips=sns.load_dataset('tips')
sns.boxplot(
    data=tips,
    x='day',
    y='total_bill',
    hue='sex'
)
```

```python
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='sex',
    style='smoker'
    
)
plt.title('tips by gender')
plt.tight_layout()
```

```python
tips.head()
pivot_tips=tips.pivot_table(index='day',columns='time',values='tip',aggfunc='mean')
sns.heatmap(
    data=pivot_tips,
    annot=True,
    
)

# day = Sun, time = Dinner
# there are many tip values.

# 👉 Pandas asks:

# “Which one should I put in the cell?”

# pivot() has no aggregation logic.
```
