```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
page_count=1
while True:
    URL=f'https://quotes.toscrape.com/page/{page_count}/'
    res=requests.get(URL)
    soup=BeautifulSoup(res.text,'lxml')
    quotes=soup.select('div.quote')
    if not quotes:
        print('no page found')
        break
    with open(f'scraped_folder/quotespage{page_count}.html','w',encoding='utf-8') as f:
        f.write(res.text)
        print(f'page {page_count} downloaded successfully')
    page_count+=1
```

```python
result=[]
for quote_page in range(1,11):
    with open(f'scraped_folder/quotespage{quote_page}.html','r',encoding='utf-8') as f:
        html_content=f.read()
    soup=BeautifulSoup(html_content,'lxml')
    
    all_quotes=soup.select('div.quote')
    for quote in all_quotes:
        tag_for_quote=[]
        all_tags=quote.select('.tags .tag')  #these is space between both classes
        for tag in all_tags:
            tag=tag.get_text()
            tag_for_quote.append(tag)
        if 'life' in tag_for_quote:
            text=quote.select('span.text')[0].get_text()
            name=quote.select_one('span .author').get_text()
            result.append([text,name])

quote_data=pd.DataFrame(result,columns=['quote','author'])
quote_data
quote_data.to_csv('cleaned_folder/quote_data.csv',index=False)

```
