```python
df=pd.read_csv('raw_data.csv')
df.drop_duplicates(inplace=True)
df.fillna(0,inplace=True)
df.sort_values('income', inplace=True)
df.to_csv('sorted_raw_data1.csv',index=False)
```

```python
df.groupby('country')['income'].mean()
df.groupby('country')['income'].agg(['mean','max','min'])
df.groupby('country').agg({
    'age':'mean',
    'gender':'count'
})
df.groupby('country')['income'].agg(avg_salary='mean',maximum_salary='max')
df.groupby('country').agg(
    avg_salary=('income','mean'),
    min_age=('age','min')
)
```

```python
df_customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Adam", "Bob", "Charlie", "Dave"]
})

df_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [2, 1, 4, 5],
    "amount": [250, 120, 300, 180]
})

pd.merge(df_customers,df_orders,on='customer_id')
pd.merge(df_customers,df_orders,on ='customer_id',how='right')
```
