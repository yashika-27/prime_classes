```python
movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232]  # in $M
plt.bar(years,revenue,color='black')
for i in range(len(revenue)):
    plt.text(years[i],revenue[i]+20,movies[i],ha='center')
plt.tight_layout()
plt.ylim(0,max(revenue)+100)
```

```python
import numpy as np
oscar_movies = ["The Dark Knight","The Hurt Locker","The King's Speech","The Artist","Argo"]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M
width=0.4

years = [2008, 2009, 2010, 2011, 2012]
#Matplotlib bar charts do NOT require “real x-axis data”. They require numeric positions.
# np.arange(len(years)) creates those numeric positions, and years are only labels, not positions.
x=np.arange(len(years))   #why we use this x because in bar graph x axis has labels only not numericals so we use real numerical values
plt.bar(x-width/2,oscar_revenue,width,label='oscar_movies')
plt.bar(x+width/2,non_oscar_revenue,width,label='non_oscar_movies',alpha=0.5)
plt.legend()
plt.xticks(x,years)
```

```python
people = ["Person A", "Person B", "Person C", "Person D", "Person E", 
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]
size=blood_pressure
colors=['red' if bp>130 else 'green' for bp in blood_pressure]
plt.scatter(age,blood_pressure,color=colors)
# plt.scatter(age,blood_pressure,s=size,c=blood_pressure,cmap='plasma',marker='<',alpha=0.5)
plt.colorbar(label='BP')
for i in range(len(people)):
    plt.annotate(people[i],xy=(age[i],blood_pressure[i]))
plt.title('Age Vs BP')
plt.xlabel('Age')
plt.ylabel('Bp')
plt.tight_layout()
plt.xlim(min(age),max(age)+10)
```

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]
explode=[0,0,0,0.1,0]
colors=['green','blue','red','purple','yellow']
plt.pie(amounts,labels=expenses,autopct='%1.1f%%',colors=colors,explode=explode,shadow=True,startangle=90,
       wedgeprops={'linestyle':'-','linewidth':2,'fill':True,'edgecolor':'black'}
       )
plt.title('expenses')
plt.tight_layout()
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22, 23, 21, 24, 25],   # New York
    [18, 19, 17, 20, 21],   # London
    [30, 32, 31, 33, 34],   # Delhi
    [25, 26, 24, 27, 28]    # Tokyo
]
fig,axes=plt.subplots(2,2)
# axes=axes.flatten()
# for i,ax in enumerate(axes):
#     ax.plot(days,temperatures[i],marker='o')
#     ax.set_title(cities[i])  #use set as this is object oriented ploting
#     ax.grid(True)

#OR

count=0
for i in range(2):
    for j in range(2):
        axes[i][j].plot(days,temperatures[count],marker='o')
        axes[i][j].set_title(cities[count])
        axes[i][j].grid(True)
    count+=1
    

fig.suptitle('Temperature over cities')  #use sup 
fig.supxlabel('Days')
fig.supylabel('Temperature (celcius)')
fig.tight_layout()
```
