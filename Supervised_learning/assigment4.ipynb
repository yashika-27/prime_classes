```python
import json
```

```python
def load_data(fileName):
    with open(fileName,'r') as f:
        data=json.load(f)
        return data
```

```python
data=load_data('interview_style_data.json')
for user in data:
    print(user)
```

```python
def clean_data(data):
    seen_names=set()
    unique_data=[]
    text_to_num={'one':1,'two':2,'three':3,'four':4,'five':5}
    for user in data:
        raw_rating=str(user['rating']).strip().lower()
        if raw_rating in text_to_num:
            raw_rating=text_to_num[raw_rating]
        user['rating']=float(raw_rating)
        if user['name'] in seen_names:
            continue
        else:
            seen_names.add(user['name'])
            unique_data.append(user)
    return unique_data
```

```python
data=clean_data(data)
print(data)
    
```

```python
def data_insights(data):
    brand1=0
    brand2=0
    brand3=0
    rating_brand1=0
    rating_brand2=0
    rating_brand3=0
    for user in data:
        match user['category']:
            case 'Laptop':
                brand1+=1
                rating_brand1+=user['rating']
            case 'Mobile':
                brand2+=1
                rating_brand2+=user['rating']
            case 'Tablet':
                brand3+=1
                rating_brand3+=user['rating']
    print(f"avg rating of laptop is : {rating_brand1/brand1}")
    print(f"avg rating of mobile is : {rating_brand2/brand2}")
    print(f"avg rating of tablet is : {rating_brand3/brand3}")
```

```python
data_insights(data)
```

```python
def pricing(data):
    brand1_price=0
    brand2_price=0
    brand3_price=0
    for user in data:
        if user['category']=='Laptop':
            if int(user['price'])>brand1_price:
                brand1_price=int(user['price'])
        if user['category']=='Mobile':
            if int(user['price'])>brand2_price:
                brand2_price=int(user['price'])
        if user['category']=='Tablet':
            if int(user['price'])>brand3_price:
                brand3_price=int(user['price'])
    print(f"highest price if laptop is : Rs {brand1_price}")
    print(f"highest price if mobile is : Rs {brand2_price}")
    print(f"highest price if tablet is : Rs {brand3_price}")
```

```python
pricing(data)
```
