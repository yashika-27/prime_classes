```python
import requests
from bs4 import BeautifulSoup
```

```python
URL='https://www.scrapethissite.com/pages/simple/'
res=requests.get(URL)
print(res.status_code)
```

```python
scraped_data=res.text
with open("scraped_folder/data.html","w", encoding="utf-8") as f:  #use encoding so that python dont crash
    f.write(scraped_data)
```

```python
with open('scraped_folder/data.html','r') as f:
    html_content=f.read()
soup=BeautifulSoup(html_content,'lxml')
soup.find('h3')
data=[]
all_h3=soup.find_all('h3')
for h3 in all_h3:
    country=h3.get_text(strip=True)
    
    population=h3.find_next("div").select("span.country-population")[0].get_text(strip=True)

    data.append([country,population])

    
```

```python
import pandas as pd
df=pd.DataFrame(data,columns=['Country','Population'])
df
```
